```python
#components of python:
# literals- values that are directly written in the code (e.g., 5 (numeric), "hello" (string), True (boolean))
# variables- names that refer to values (e.g., x = 5)
# operators- symbols that perform operations on values (e.g., +, -, *, /)
# statements- instructions that the Python interpreter can execute (e.g., print("Hello, World!"))
# expressions- combinations of variables, literals, and operators that evaluate to a single value (e.g., 5 + 3)
```

```python
#primitive data types in python:
#int - integer data type, used to represent whole numbers (e.g., 1, 2, 3)
#float - floating-point data type, used to represent decimal numbers (e.g., 1
#.0, 2.5, 3.14)
#str - string data type, used to represent text (e.g., "hello", 'world')
#bool - boolean data type, used to represent truth values (e.g., True, False)
#None - special data type, used to represent the absence of a value (e.g., None)
```
