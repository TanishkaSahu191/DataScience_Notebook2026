```python
#strings
s1="hello world"
print(type(s1))

print(len(s1)) #length of the string
#space is also counted as a character in the string, so the length of "hello world" is 11

#indexing
#positioning of characters in a string, starting from 0 for the first character, 1 for the second character, and so on. Negative indexing starts from -1 for the last character, -2 for the second last character, and so on.
print(s1[0])
print(s1[1])
print(s1[-1]) #last character
print(s1[-2]) #second last character
```

```python
#type casting - converting one data type to another

num1=100
print(num1)
print(type(num1))
float(num1)

num1=float(num1) #this will convert the value of num1 to float and assign it to num1
print(num1)
#num1 will not change to float because we did not assign the float value to the variable, we just converted it to float and printed it. To change the value of num1 to float, we need to assign the float value to num1 like this: num1 = float(num1)
print(type(num1)) #this will now print <class 'float'> because we assigned the float value to the variable
```

```python
#operators in python: symbols to perform certain operations
#arithmetic operators - used to perform mathematical operations on numerical values (+, -, *, /,%,//)
x=10
y=x+20
print(x)
print(y)

#x and y are variables(operands) that store numerical values. The + operator is used to add the values of x and 20, and the result is stored in the variable y. The print() function is used to display the values of x and y on the console.
```

```python
#floor division operator (//) - used to perform division and return the largest integer less than or equal to the result of the division
a=100
b=12
c=a//b
print(c) #this will print 8 because 100 divided by 12 is 8.3333 and the largest integer less than or equal to 8.3333 is 8
```

```python
#exponnentiation operator (**) - used to perform exponentiation, which is raising a number to the power of another number
print(5**2)
```

```python
#assignment operator (=) - used to assign a value to a variable
x=10 #this will assign the value 10 to the variable x
```

```python
#compound assignment operators - used to perform an operation and assign the result to a variable in a single step (e.g., +=, -=, *=, /=)
x=10
x+=2
print(x)
```

```python
#comparison operators - used to compare two values and return a boolean value (True or False) based on the comparison (==, !=, >, <, >=, <=)
a=10
b=20

print(a==b)
print(a>b)

print("python"=="python")
print("python"!="python")
print("python"=="Python") #case sensitive
```

```python
#logical operators - used to combine multiple boolean expressions and return a boolean value based on the logical operation (and, or, not)
x=True
y=False
print(x and y) #this will print False because both x and y are not True
print(x or y) #this will print True because at least one of x or y is True
print(not x) #this will print False because x is True
```

```python
#associativity of operators - determines the order in which operators of the same precedence are evaluated in an expression. In Python, most operators have left-to-right associativity, meaning that they are evaluated from left to right. However, some operators, such as the exponentiation operator (**), have right-to-left associativity, meaning that they are evaluated from right to left.

#precedence of operators - determines the order in which operators are evaluated in an expression. Operators with higher precedence are evaluated before operators with lower precedence. For example, the multiplication operator (*) has higher precedence than the addition operator (+), so in the expression 2 + 3 * 4, the multiplication is performed first, resulting in 2 + 12 = 14.

```

```python
#coding exercises

#area of triangle

a=float(input("first side:"))
b=float(input("second side:"))
c=float(input("third side:"))
s=(a+b+c)/2

#s is semi perimeter

x=s*(s-a)*(s-b)*(s-c)
area=(pow(x,0.5))

print("area of triangle is: ",round(area,2))
```
